```python
from sklearn.datasets import fetch_openml

def load_mnist():
    mnist = fetch_openml("mnist_784", as_frame=False)
    X, y = mnist.data / 255.0, mnist.target
    X_train, X_test = X[:60000], X[60000:]
    y_train, y_test = y[:60000], y[60000:]
    y_train_5 = (y_train == "5").astype(float).reshape(-1, 1)
    y_test_5 = (y_test == "5").astype(float).reshape(-1, 1)
    return X_train, X_test, y_train_5, y_test_5
```

```python
X_train, X_test, y_train_5, y_test_5 = load_mnist()

X_b = np.c_[np.ones((len(X_train),1)), X_train]
X_b_test = np.c_[np.ones((len(X_test), 1)),  X_test] 
```

```python
class LinearRegression:
    def __init__(self):
       self.theta = None
        
    def fit(self, X, y):
        X_b = np.c_[np.ones((len(X),1)), X]
        self.theta = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
        return self

    def predict(self, X_new):
       X_new_b = np.c_[np.ones((len(X_new), 1)), X_new]
       return X_new_b @ self.theta         
    
```

```python
linear_regression = LinearRegression()
linear_regression.fit(X_train, y_train_5)
theta_vector = linear_regression.predict(X_test)
theta_vector
```
